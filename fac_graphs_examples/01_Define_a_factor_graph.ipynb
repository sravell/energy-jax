## Factor Graphs

Factor graphs are used to represent factorizable scalar functions

$$ g(x_1, x_2, \dots, x_n) = \prod_i f_i(X_i)$$

Where each factor $f_i$ is a function of a subset of the $x_j$, $X_i \subset \{x_1, \dots, x_n\}$.

To define a graph, we have to define the variables $x_j$ and the factors $f_i$. 



## Variables

Each variable is essentially a placeholder for a tensor. Therefore, to define a Variable, we have to define the shape of tensor it represents. A variable can also either be continuous or discrete. You can also asign a name to Variables for visualization purposes.

```python
variables = [VariableNode(name="V{}".format(i)) for i in range(5)]
```

## Factors

A factor is composed of a set of Variables and a function that takes in Tensors of the shapes defined by its variables and outputs a scalar

```python
factor_1 = FactorNode([variables[0], variables[1]], lambda x1, x2: 1, name="F1")
factor_2 = FactorNode(
    [variables[1], variables[2], variables[3]], lambda x1, x2, x3: 1, name="F2"
)
```

## Putting it all Together

we can then construct a FactorGraph out of these factors

```python
fg = FactorGraph([factor_1, factor_2])
```

```python
draw_graph(fg, figsize=(4, 4))
```
